## 1. Functions

A **function** is a reusable, named block of code that performs a task. Instead of copy-pasting
the same logic everywhere, we write it once inside a function and _call_ it whenever we need it.

### 1.1 Defining functions with `def`


````python
# The simplest possible function: no inputs, no outputs, just an action
def greet():
    # everything indented under 'def' belongs to the function body
    print("Hello! Welcome to Day 3 of Python.")

# Defining a function does NOT run it. We must CALL it using ()
greet()
greet()   # functions can be called as many times as we like
````

````python
# A function becomes useful when it can accept input.
# 'name' here is called a PARAMETER (a placeholder defined in the function signature)
def greet_person(name):
    print(f"Hello, {name}! Good morning.")

# When we call the function, "Netra" is the ARGUMENT (the actual value passed in)
greet_person("Netra")
greet_person("Aarav")
````

### 1.2 Parameters vs Arguments

- **Parameter**: the variable name listed inside the parentheses in the function _definition_.
- **Argument**: the actual _value_ you pass in when _calling_ the function.

```python
def greet_person(name):   # 'name' is a PARAMETER
    print(name)

greet_person("Netra")     # "Netra" is an ARGUMENT
```

Let's see a function with multiple parameters/arguments.


````python
def add_numbers(a, b):          # a, b are PARAMETERS
    result = a + b
    print(f"{a} + {b} = {result}")

add_numbers(5, 3)                # 5, 3 are ARGUMENTS (matched by position -> a=5, b=3)
add_numbers(10, 20)
````

### 1.3 Default Arguments

We can give a parameter a **default value**. If the caller doesn't provide that argument,
the default is used automatically. Default parameters must come _after_ non-default ones.


````python
# 'greeting' has a default value of "Hello"
def greet_with_message(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet_with_message("Netra")                  # uses the default greeting
greet_with_message("Netra", "Good evening")  # overrides the default
````

````python
# Medium example: default arguments in a small "billing" calculation
def calculate_total(price, quantity=1, discount_percent=0):
    # discount_percent defaults to 0 if not supplied
    subtotal = price * quantity
    discount_amount = subtotal * (discount_percent / 100)
    total = subtotal - discount_amount
    return total

print(calculate_total(100))                     # quantity=1, no discount
print(calculate_total(100, 3))                   # 3 items, no discount
print(calculate_total(100, 3, discount_percent=10))  # 3 items with 10% discount
````

### 1.4 Keyword Arguments

Arguments can be passed by **position** (order matters) or by **keyword** (name=value, order
doesn't matter). Keyword arguments make code more readable, especially with many parameters.


````python
def describe_student(name, course, batch):
    print(f"{name} is enrolled in {course}, Batch {batch}")

# Positional call: order MUST match the function definition
describe_student("Netra", "Data Science & AI/ML", "2026-A")

# Keyword call: order does NOT matter because we name each argument
describe_student(batch="2026-A", name="Netra", course="Data Science & AI/ML")

# Mixed: positional arguments must always come BEFORE keyword arguments
describe_student("Netra", batch="2026-A", course="Data Science & AI/ML")
````

### 1.5 `*args` and `**kwargs`

Sometimes we don't know in advance _how many_ arguments will be passed.

- `*args` collects any number of extra **positional** arguments into a **tuple**.
- `**kwargs` collects any number of extra **keyword** arguments into a **dictionary**.

(The names `args` and `kwargs` are just convention — the `*` and `**` are what matter.)


````python
# *args example: sum any number of numbers
def add_all(*args):
    print(f"args received as a tuple: {args}")
    return sum(args)

print(add_all(1, 2))
print(add_all(1, 2, 3, 4, 5))
print(add_all())   # works even with zero arguments -> empty tuple, sum = 0
````

````python
# **kwargs example: print any number of named details
def print_profile(**kwargs):
    print(f"kwargs received as a dict: {kwargs}")
    for key, value in kwargs.items():
        print(f"  {key} -> {value}")

print_profile(name="Netra", role="Backend/ML Engineer", company="Zakipoint Health")
````

````python
# Medium example: combining normal params, *args, and **kwargs together
# Order in the signature MUST be: normal params -> *args -> default params -> **kwargs
def build_report(title, *sections, author="Unknown", **extra_info):
    print(f"REPORT: {title}")
    print(f"Author: {author}")
    print("Sections:", sections)
    print("Extra info:", extra_info)

build_report(
    "Day 3 Class Summary",
    "Functions", "Modules", "Practice Exercises",   # collected into *sections
    author="Netra",
    duration="2 hours",
    level="Basic to Medium"                          # collected into **extra_info
)
````

### 1.6 Return Values and Multiple Return Values

- `return` sends a value back to the caller and immediately ends the function.
- A function with no `return` statement returns `None` by default.
- Python lets you return **multiple values** — they are packed into a tuple automatically.


````python
def square(n):
    return n * n

result = square(6)
print("Square of 6 is:", result)

# A function without 'return' gives back None
def log_message(msg):
    print(f"[LOG] {msg}")
    # no return statement here

output = log_message("System started")
print("Return value of log_message:", output)   # None
````

````python
# Returning multiple values: Python packs them into a tuple behind the scenes
def get_min_max(numbers):
    return min(numbers), max(numbers)   # two values returned

scores = [72, 88, 91, 65, 100, 59]

# We can "unpack" the returned tuple directly into two variables
lowest, highest = get_min_max(scores)
print(f"Lowest score: {lowest}, Highest score: {highest}")

# Or keep it as a single tuple if we prefer
result_tuple = get_min_max(scores)
print("As a tuple:", result_tuple)
````

````python
# Medium example: a function that validates AND computes, returning multiple pieces of info
def analyze_marks(marks):
    if not marks:
        return False, 0, "No marks provided"   # multiple values even in an edge case

    average = sum(marks) / len(marks)
    status = "Pass" if average >= 40 else "Fail"
    return True, round(average, 2), status

is_valid, avg, status = analyze_marks([45, 60, 38, 72])
print(f"Valid data: {is_valid}, Average: {avg}, Status: {status}")

is_valid, avg, status = analyze_marks([])
print(f"Valid data: {is_valid}, Average: {avg}, Status: {status}")
````

### 1.7 Variable Scope: Local, Global, and the `global` keyword

- A variable created **inside** a function is **local** — it only exists while that function runs
  and cannot be accessed outside it.
- A variable created **outside** any function (at the top level of the script/notebook) is
  **global** — it can be _read_ from inside functions, but not _modified_ unless you say so
  explicitly using the `global` keyword.


````python
def show_local_variable():
    local_message = "I only exist inside this function"
    print(local_message)

show_local_variable()

try:
    print(local_message)   # this will fail: local_message doesn't exist out here
except NameError as e:
    print("Error:", e)
````

````python
# Global variables CAN be read inside a function without any special keyword
course_name = "Data Science & AI/ML"   # this is a global variable

def show_course():
    # we are only READING the global variable here, which is allowed
    print(f"Current course: {course_name}")

show_course()
````

````python
# But if we try to MODIFY a global variable inside a function without 'global',
# Python creates a NEW local variable instead (a very common beginner mistake!)
counter = 0

def increment_wrong():
    counter = counter + 1   # this actually raises an error, because Python treats
    print(counter)          # 'counter' as local the moment we assign to it, before it's defined

try:
    increment_wrong()
except UnboundLocalError as e:
    print("Error:", e)
````

````python
# The 'global' keyword tells Python: "use the global variable, don't create a local one"
counter = 0

def increment_correct():
    global counter          # explicitly declare we mean the global 'counter'
    counter = counter + 1
    print("Counter is now:", counter)

increment_correct()
increment_correct()
increment_correct()
print("Final counter value:", counter)
````

### 1.8 Lambda Functions

A `lambda` is a small, **anonymous** (unnamed) function written in a single line. It's useful
for short, throwaway logic — especially as an argument to other functions like `sorted()`,
`map()`, or `filter()` (coming up next).

Syntax: `lambda parameters: expression` — the expression's value is automatically returned.


````python
# A regular function...
def square_fn(x):
    return x * x

# ...and the equivalent lambda function
square_lambda = lambda x: x * x

print(square_fn(5))
print(square_lambda(5))
````

````python
# Lambdas can take multiple arguments too
multiply = lambda a, b: a * b
print(multiply(4, 7))

# Medium example: using a lambda directly as a "key" function to sort custom data
students = [("Netra", 91), ("Aarav", 78), ("Priya", 85)]

# sorted() needs to know WHAT to sort by; we tell it using a lambda
sorted_by_score = sorted(students, key=lambda student: student[1], reverse=True)
print("Ranked by score (highest first):", sorted_by_score)
````

## 2. Built-in Functional Tools

Python ships with several very handy built-in functions that operate _on other functions or
collections_. These let us write short, expressive code instead of long loops.

### 2.1 `map()`

`map(function, iterable)` applies `function` to **every item** in `iterable` and returns a
map object (we usually wrap it in `list()` to see the results).


````python
numbers = [1, 2, 3, 4, 5]

# Using a named function with map()
def double(n):
    return n * 2

doubled = list(map(double, numbers))
print("Doubled:", doubled)

# The same thing, but with a lambda (very common combination: map + lambda)
tripled = list(map(lambda n: n * 3, numbers))
print("Tripled:", tripled)
````

### 2.2 `filter()`

`filter(function, iterable)` keeps only the items for which `function` returns `True`.


````python
numbers = [10, 15, 22, 33, 40, 55, 68]

# Keep only even numbers
even_numbers = list(filter(lambda n: n % 2 == 0, numbers))
print("Even numbers:", even_numbers)

# Medium example: filter a list of dictionaries based on a condition
students = [
    {"name": "Netra", "score": 91},
    {"name": "Aarav", "score": 38},
    {"name": "Priya", "score": 85},
    {"name": "Rohan", "score": 45},
]

passed_students = list(filter(lambda s: s["score"] >= 40, students))
print("Passed students:", passed_students)
````

### 2.3 `zip()`

`zip()` combines multiple iterables **element-by-element** into tuples. Very useful for
pairing up related lists.


````python
names = ["Netra", "Aarav", "Priya"]
scores = [91, 78, 85]

# zip() pairs items at the same position together
paired = list(zip(names, scores))
print("Paired:", paired)

# Common pattern: loop over the zipped pairs directly
for name, score in zip(names, scores):
    print(f"{name} scored {score}")
````

### 2.4 `enumerate()`

`enumerate()` gives you **both the index and the value** while looping — no need to manually
manage a counter variable.


````python
subjects = ["Functions", "Modules", "Standard Library"]

# Without enumerate (the old, clunky way)
i = 0
for subject in subjects:
    print(i, subject)
    i += 1

print("---")

# With enumerate (clean and Pythonic)
for index, subject in enumerate(subjects):
    print(index, subject)

print("---")

# enumerate() can also start counting from a number you choose
for index, subject in enumerate(subjects, start=1):
    print(f"Topic {index}: {subject}")
````

### 2.5 `sorted()`, `len()`, `sum()`, `min()`, `max()`

A quick tour of the everyday built-ins you'll use constantly.


````python
scores = [72, 91, 45, 88, 60, 100, 38]

print("Original list:", scores)
print("Length      :", len(scores))
print("Sum         :", sum(scores))
print("Minimum     :", min(scores))
print("Maximum     :", max(scores))
print("Sorted (asc):", sorted(scores))
print("Sorted (desc):", sorted(scores, reverse=True))
````

````python
# Medium example: sorted() with a custom key, combined with everything above
students = [
    {"name": "Netra", "score": 91},
    {"name": "Aarav", "score": 38},
    {"name": "Priya", "score": 85},
    {"name": "Rohan", "score": 45},
]

# Sort students by score, highest first
ranked = sorted(students, key=lambda s: s["score"], reverse=True)

print("Class Ranking:")
for position, student in enumerate(ranked, start=1):
    print(f"  {position}. {student['name']} - {student['score']}")

print()
all_scores = [s["score"] for s in students]   # a quick list comprehension
print(f"Class average: {sum(all_scores) / len(all_scores):.2f}")
print(f"Topper score  : {max(all_scores)}")
print(f"Lowest score  : {min(all_scores)}")
````

## 3. Modules and Packages

A **module** is simply a `.py` file containing Python code (functions, variables, classes) that
you can reuse in other files. A **package** is a folder of related modules. Splitting code into
modules keeps large projects organized and avoids rewriting the same logic everywhere.

### 3.1 `import`, `from ... import`, aliasing with `as`


````python
# Import an entire module, then access its contents using dot notation
import math

print(math.pi)              # a constant defined inside the math module
print(math.sqrt(64))        # a function defined inside the math module
````

````python
# Import ONLY specific names from a module, so you can use them directly (no dot notation)
from math import sqrt, pi

print(sqrt(81))
print(pi)
````

````python
# Aliasing a module with 'as' gives it a shorter/conventional name
import math as m

print(m.floor(9.7))   # rounds down
print(m.ceil(9.2))    # rounds up

# Aliasing is extremely common with data libraries, e.g.:
#   import numpy as np
#   import pandas as pd
````

### 3.2 Standard Library Tour: `math`, `random`, `datetime`, `os`, `sys`

Python comes with a huge "standard library" of modules that are always available — no
installation needed. Let's explore five of the most commonly used ones.


**`math`** — mathematical functions and constants


````python
import math

print("pi            :", math.pi)
print("e             :", math.e)
print("sqrt(2)       :", math.sqrt(2))
print("2 raised to 10:", math.pow(2, 10))
print("factorial(5)  :", math.factorial(5))
print("ceil(4.3)     :", math.ceil(4.3))
print("floor(4.7)    :", math.floor(4.7))
````

**`random`** — generating random numbers and choices


````python
import random

print("Random float between 0 and 1:", random.random())
print("Random int between 1 and 6 (dice roll):", random.randint(1, 6))

fruits = ["apple", "banana", "cherry", "mango"]
print("Random choice from a list:", random.choice(fruits))

# shuffle() rearranges a list in place (modifies the original list)
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print("Shuffled list:", numbers)

# sample() picks multiple UNIQUE random items without repetition
print("Random sample of 2 fruits:", random.sample(fruits, 2))
````

**`datetime`** — working with dates and times


````python
import datetime

# Current date and time
now = datetime.datetime.now()
print("Right now:", now)

# Just today's date
today = datetime.date.today()
print("Today's date:", today)

# Creating a specific date
birthday = datetime.date(2000, 5, 15)
print("A specific date:", birthday)

# Date arithmetic: how many days until a future date?
future_date = datetime.date(2026, 12, 25)
days_left = (future_date - today).days
print(f"Days left until {future_date}: {days_left}")

# Formatting a date as a readable string
print("Formatted:", now.strftime("%d-%m-%Y %H:%M:%S"))
````

### 3.3 Introduction to Virtual Environments (`venv`) and `pip`

These commands are run in your **terminal / command prompt**, not inside a notebook cell
(they're shown here as text so you can copy them into your terminal later).

**Why virtual environments?**
Different projects often need different, sometimes conflicting, versions of the same package.
A **virtual environment** is an isolated Python installation just for one project, so packages
installed for Project A never clash with Project B.

```bash
# 1. Create a virtual environment named 'venv' in your project folder
python -m venv venv

# 2. Activate it
#    On Windows:
venv\Scripts\activate
#    On macOS/Linux:
source venv/bin/activate

# Once activated, your terminal prompt usually shows (venv) at the start

# 3. Install packages using pip (Python's package installer) — installs ONLY into this venv
pip install requests
pip install pandas numpy

# 4. See what's installed in the current environment
pip list

# 5. Save your project's exact dependencies to a file (for sharing / reproducing the setup)
pip freeze > requirements.txt

# 6. Someone else (or you, on another machine) can recreate the same environment:
pip install -r requirements.txt

# 7. Deactivate the environment when you're done
deactivate
```

**Rule of thumb:** one virtual environment per project. Never install project-specific
packages into your "global" system Python.


### 3.4 Organizing Code Into Reusable Modules

In real projects, we don't write everything in one giant file. We split related functions into
their own `.py` files (modules), then `import` them wherever needed.

Below, we'll simulate this inside the notebook by **writing a `.py` file to disk**, then
importing it — exactly how it would work in a real project across multiple files.


````python
# Step 1: Write a small module to disk called 'calculator_utils.py'
# In a real project, this would just be a separate file you create in your editor.

module_code = '''
# calculator_utils.py
# A small reusable module with basic math helper functions.

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero")
    return a / b

# A module-level constant, usable after import
VERSION = "1.0"
'''

with open("calculator_utils.py", "w") as f:
    f.write(module_code)

print("calculator_utils.py has been created.")
````

````python
# Step 2: Import and use our own custom module, just like math or random!
import calculator_utils

print("Module version:", calculator_utils.VERSION)
print("5 + 3 =", calculator_utils.add(5, 3))
print("10 - 4 =", calculator_utils.subtract(10, 4))
print("6 * 7 =", calculator_utils.multiply(6, 7))
print("20 / 4 =", calculator_utils.divide(20, 4))
````

````python
# We can also use 'from ... import' and aliasing with OUR OWN module, same as any other
from calculator_utils import add as sum_two_numbers

print(sum_two_numbers(100, 250))
````

**Real-project convention:** a typical package folder looks like this:

```
my_project/
│
├── main.py                 # entry point, imports from the modules below
├── calculator_utils.py     # a module of related helper functions
├── string_utils.py         # another module, e.g. text-processing helpers
└── requirements.txt        # list of pip packages this project depends on
```

Any folder containing an `__init__.py` file (can be empty) is treated as a **package**,
letting you group related modules together, e.g. `from my_project.utils import calculator_utils`.

## Recap

| Topic                                     | Key takeaway                                                                 |
| ----------------------------------------- | ---------------------------------------------------------------------------- |
| `def` functions                           | Reusable blocks of code, defined once, called many times                     |
| Parameters vs arguments                   | Parameters = placeholders in definition; arguments = actual values passed    |
| Default / keyword arguments               | Make functions flexible and calls more readable                              |
| `*args` / `**kwargs`                      | Accept a variable number of positional / keyword arguments                   |
| Return values                             | `return` sends data back; multiple values are returned as a tuple            |
| Scope & `global`                          | Local variables live inside functions; `global` lets you modify a global one |
| `lambda`                                  | Small anonymous one-line functions, often used with `sorted`/`map`/`filter`  |
| `map`, `filter`, `zip`, `enumerate`       | Powerful one-line alternatives to manual loops                               |
| `sorted`, `len`, `sum`, `min`, `max`      | Everyday built-ins for working with collections                              |
| `import` / `from...import` / `as`         | Ways to bring external code into your file                                   |
| `math`, `random`, `datetime`, `os`, `sys` | Commonly used standard library modules                                       |
| `venv` + `pip`                            | Isolate project dependencies; install/manage packages                        |
| Custom modules                            | Split code across `.py` files and `import` them for reuse                    |
